**Q6. Perform replacing missing values with KNN imputers on scaled data. Also, inverse the scaled data to get the original data.**


```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler

df = house_data.copy()

df.loc[0, 'Price'] = np.nan

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[numeric_cols])
imputed_data = KNNImputer(n_neighbors=5).fit_transform(scaled_data)
imputed_data = pd.DataFrame(scaler.inverse_transform(imputed_data), columns=numeric_cols)

print("First 5 rows of the DataFrame after KNN imputation and inverse scaling:")
print(imputed_data.head().to_markdown(index=False, numalign="left", stralign="left"))
print("\nMissing values after KNN imputation:")
print(imputed_data.isnull().sum().to_markdown())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
KOL_house_data_missing.columns
```

Cleaning the columns

```python
def clean_price(price):
    if isinstance(price, str):
        price = price.replace('₹', '').replace(',', '')
        try:
            if 'Cr' in price:
                return float(price.replace('Cr', '').strip()) * 10000000
            elif 'L' in price:
                return float(price.replace('L', '').strip()) * 100000
            else:
                return float(price.strip())
        except ValueError:
            return np.nan # Return NaN for values that cannot be converted
    return price

def clean_emi(emi):
    if isinstance(emi, str):
        emi = emi.replace('₹', '').replace(',', '')
        try:
            if 'Lac' in emi:
                return float(emi.replace('Lac', '').strip()) * 100000
            elif 'K' in emi:
                return float(emi.replace('K', '').strip()) * 1000
            else:
                return float(emi.strip())
        except ValueError:
            return np.nan # Return NaN for values that cannot be converted
    return emi

def clean_sqft(sqft):
    if isinstance(sqft, str):
        sqft = sqft.replace('sq.ft', '').replace(',', '').strip()
        try:
            return float(sqft)
        except ValueError:
            return np.nan # Return NaN for values that cannot be converted
    return sqft

def clean_price_per_sqft(price_per_sqft):
    if isinstance(price_per_sqft, str):
        price_per_sqft = price_per_sqft.replace('₹', '').replace(',', '').replace('/sq.ft', '').strip()
        try:
            if 'K' in price_per_sqft:
                return float(price_per_sqft.replace('K', '').strip()) * 1000
            else:
                return float(price_per_sqft.strip())
        except ValueError:
            return np.nan # Return NaN for values that cannot be converted
    return price_per_sqft

def clean_bhk(bhk):
    if isinstance(bhk, str):
        bhk = bhk.replace(' BHK', '').strip()
        try:
            return float(bhk)
        except ValueError:
            return np.nan
    return bhk


KOL_house_data_missing['Flat_Price'] = KOL_house_data_missing['Flat_Price'].apply(clean_price)
KOL_house_data_missing['EMI_Starts'] = KOL_house_data_missing['EMI_Starts'].apply(clean_emi)
KOL_house_data_missing['BHK'] = KOL_house_data_missing['BHK'].apply(clean_bhk)
KOL_house_data_missing['Total_Sq.ft'] = KOL_house_data_missing['Total_Sq.ft'].apply(clean_sqft)
KOL_house_data_missing['Price_per_sq.ft'] = KOL_house_data_missing['Price_per_sq.ft'].apply(clean_price_per_sqft)

display(KOL_house_data_missing.head())
```

Splitting the data into training and testing sets (80% train, 20% test)

```python
X_kol = imputed_data_kol[['Total_Sq.ft', 'Price_per_sq.ft']]
y_kol = imputed_data_kol['Flat_Price']
```

```python
from sklearn.model_selection import train_test_split

X_train_kol, X_test_kol, y_train_kol, y_test_kol = train_test_split(X_kol, y_kol, test_size=0.4, random_state=123)

print("Shape of X_train_kol:", X_train_kol.shape)
print("Shape of X_test_kol:", X_test_kol.shape)
print("Shape of y_train_kol:", y_train_kol.shape)
print("Shape of y_test_kol:", y_test_kol.shape)
```

Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
LR_kol = LinearRegression()
LR_model_kol = LR_kol.fit(X_train_kol, y_train_kol)
y_pred_kol = LR_model_kol.predict(X_test_kol)
```

```python
MSE_kol = mean_squared_error(y_test_kol, y_pred_kol)
R_square_kol = r2_score(y_test_kol, y_pred_kol)
print("Mean Squared Error (MSE):", MSE_kol)
print("R-squared (R2):", R_square_kol)
```

Decision Tree Regression

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score

dt_regressor = DecisionTreeRegressor(random_state=123)
dt_model = dt_regressor.fit(X_train_kol, y_train_kol)
y_pred_dt = dt_model.predict(X_test_kol)

mse_dt = mean_squared_error(y_test_kol, y_pred_dt)
r2_dt = r2_score(y_test_kol, y_pred_dt)

print("Mean Squared Error (MSE):", mse_dt)
print("R-squared (R2):", r2_dt)
```

Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

rf_regressor = RandomForestRegressor(random_state=123)
rf_model = rf_regressor.fit(X_train_kol, y_train_kol)
y_pred_rf = rf_model.predict(X_test_kol)

mse_rf = mean_squared_error(y_test_kol, y_pred_rf)
r2_rf = r2_score(y_test_kol, y_pred_rf)

print("Mean Squared Error (MSE):", mse_rf)
print("R-squared (R2):", r2_rf)
```
